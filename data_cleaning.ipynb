```python
import pandas as pd

# Load Chicago and New Orleans data
chicago_df = pd.read_csv('/content/listings.csv')
new_orleans_df = pd.read_csv('/content/listings (1).csv')
```

```python
# Concatenate the two datasets into one
combined_df = pd.concat([chicago_df, new_orleans_df], ignore_index=True)
```

```python
combined_df.isnull().sum()
```

```python
combined_df.head()
```

```python
combined_df.shape
```

```python
combined_df['price'] = combined_df['price'].fillna(combined_df['price'].median())
```

```python
combined_df['last_review'] = combined_df['last_review'].fillna('No Reviews')
```

```python
combined_df['reviews_per_month'] = combined_df['reviews_per_month'].fillna(0)
```

```python
combined_df['license'] = combined_df['license'].fillna('Not Available')
```

```python
# Drop the 'neighbourhood_group' column
combined_df = combined_df.drop(columns=['neighbourhood_group'])

# Check if the column was removed
print(combined_df.head())
```

```python
# Check data types
print(combined_df.dtypes)


```

```python


# Convert columns to correct data types if needed
combined_df['last_review'] = pd.to_datetime(combined_df['last_review'], errors='coerce')
combined_df['price'] = combined_df['price'].astype(float)
```

```python
print(combined_df.dtypes)
```

```python
# Check price distribution
print(combined_df['price'].describe())
```

```python
combined_df['room_type'] = combined_df['room_type'].astype('category')
combined_df['host_name'] = combined_df['host_name'].astype('category')
```

```python
combined_df = combined_df.drop_duplicates()
```

```python
# Save the cleaned dataset
combined_df.to_csv('cleaned_airbnb_listings.csv', index=False)
```
